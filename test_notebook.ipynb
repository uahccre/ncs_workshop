If all of this works, it should be fairly easy to create a couple of shareable notebooks for the workshop. 

## Creating a Gemini API Key
1. Navigate to [Google AI Studio](https://aistudio.google.com)
2. Accept the ToS if required 
3. Click the "Get API Key" button (bottom left of the screen that looks like a key)
4. Click "Create API Key" in the top right 
5. Create a new project. I named mine "NCS Workshop" and the key name was "Gemini API Key" 
6. Copy your AI key to a text file for adding to Colab from the popup window

## Making a Copy of the Notebook
1. Navigate to the [UAH CCRE GitHub](https://github.com/uahccre/ncs_workshop). (You should already be here if you're reading this. This will go in the handout we send to the attendees). 
2. Click the "Open in Colab" Button
3. The file will open in Read Only mode. Go to File -> Save a Copy in Drive
4. A new tab will open with your editable copy for the rest of the lab. 

## Adding the Gemini Key to Colab Secrets
1. In your notebook copy, click the key icon on the far left sidebar that will open the secrets panel
2. Click "Add New Secret" 
3. For Name type exactly `GEMINI_API_KEY`
4. For Value, paste your key you copied above
5. Turn on the Notebook access toggle next to the secret
