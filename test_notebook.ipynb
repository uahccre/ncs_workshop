## Verifying Colab Can Access Gemini and Smolagents

```python
# Install smolagents with the LiteLLM backend required for Gemini
!pip install -q "smolagents[litellm]"
```

```python
# Load the Gemini API key from Colab Secrets
import os
try:
    from google.colab import userdata
    GEMINI_API_KEY = userdata.get("GEMINI_API_KEY")
except Exception as e:
    raise RuntimeError(
        "Couldn't get GEMINI_API_KEY from Colab Secrets. Open the secrets panel on the left sidebar: '\n"
        "Add a secret named GEMINI_API_KEY with your copied API key value and turn on notebook access.\n"
        "Then try running the code cell again."
    )

if not GEMINI_API_KEY:
    raise RuntimeError("GEMINI_API_KEY is empty. Add it in the secrets panel and make sure the name is exactly GEMINI_API_KEY and \n"
                       "verify that notebook access is on. Then rerun the code cell")

os.environ["GEMINI_API_KEY"] = GEMINI_API_KEY # LiteLLM also reads the key from here

# Build a model backed by Gemini's free tier
from smolagents import CodeAgent, LiteLLMModel

model = LiteLLMModel(model_id="gemini/gemini-3.6-flash", api_key=GEMINI_API_KEY)

# Build a simple agent
agent = CodeAgent(tools=[], model=model, max_steps=5, verbosity_level=1)

# Run a small multi-step task to confirm everything works end to end
result = agent.run("What is the sum of all prime numbers below 50? Show your reasoning.")

print("\n" + "=" * 60)
print("Success. SmolAgents and Gemini are both working")
print("Agent's answer: ", result)
print("=" * 60)
```
